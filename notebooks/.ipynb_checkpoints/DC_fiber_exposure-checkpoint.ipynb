```python
from __future__ import annotations

import h3
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt

import cartopy.crs as ccrs
import cartopy.feature as cfeature

from shapely.geometry import Polygon
from pathlib import Path


# -----------------------------
# Paths
# -----------------------------

DATA_PATH = Path(
    "/home/server/pi/homes/queenl/projects/climate-risk-dc/data"
)

FIBER_PATH = DATA_PATH / "fiber" / "bdc_41_FibertothePremises_fixed_broadband_D25_08jul2026.csv"

# -----------------------------
# Coordinate systems
# -----------------------------

CRS_GEOGRAPHIC = "EPSG:4326"


# -----------------------------
# Map settings
# -----------------------------

OREGON_EXTENT = [
    -125,   # west
    -116,   # east
    42,     # south
    46.5    # north
]

MAP_PROJECTION = ccrs.PlateCarree()


# -----------------------------
# H3 settings
# -----------------------------

FIBER_H3_INPUT_RESOLUTION = 8

# Lower = larger hexagons
FIBER_H3_PLOT_RESOLUTION = 6
```

```python
def load_fiber() -> pd.DataFrame:
    """Load fiber dataset and filter to Oregon."""

    df = pd.read_csv(FIBER_PATH)

    df = df[df["state_usps"] == "OR"]
    df = df[df["max_advertised_download_speed"] >= 1000]
    
    return df
```

```python
def plot_fiber_density(
    df: pd.DataFrame,
    resolution: int = FIBER_H3_PLOT_RESOLUTION
) -> None:
    """
    Plot fiber infrastructure density in Oregon using H3 aggregation.

    Parameters
    ----------
    df : pd.DataFrame
        Fiber locations containing h3_res8_id.
    resolution : int
        Target H3 resolution. Lower values create larger hexagons.
    """

    if df.empty:
        raise ValueError("Fiber dataframe is empty.")


    # Aggregate H3 cells to larger parent cells
    density = (
        df.assign(
            h3_id=df["h3_res8_id"].apply(
                lambda x: h3.cell_to_parent(
                    x,
                    resolution
                )
            )
        )
        .groupby("h3_id")
        .agg(
            provider_count=(
                "provider_id",
                "nunique"
            ),
            location_count=(
                "location_id",
                "count"
            )
        )
        .reset_index()
    )


    # Convert H3 cells to polygons
    density["geometry"] = density["h3_id"].apply(
        lambda h: Polygon(
            [
                (lon, lat)
                for lat, lon in h3.cell_to_boundary(h)
            ]
        )
    )


    hex_gdf = gpd.GeoDataFrame(
        density,
        geometry="geometry",
        crs=CRS_GEOGRAPHIC
    )


    # -----------------------------
    # Plot
    # -----------------------------

    fig, ax = plt.subplots(
        figsize=(8, 10),
        dpi=150,
        subplot_kw={
            "projection": MAP_PROJECTION
        }
    )


    ax.set_extent(
        OREGON_EXTENT,
        crs=MAP_PROJECTION
    )


    # Basemap
    ax.add_feature(
        cfeature.LAND,
        facecolor="whitesmoke"
    )

    ax.add_feature(
        cfeature.OCEAN,
        facecolor="lightblue"
    )

    ax.add_feature(
        cfeature.STATES,
        linewidth=0.6
    )

    ax.add_feature(
        cfeature.COASTLINE,
        linewidth=0.6
    )


    # Fiber density
    hex_gdf.plot(
        ax=ax,
        transform=MAP_PROJECTION,
        column="location_count",
        cmap="viridis",
        edgecolor="white",
        linewidth=0.15,
        legend=True
    )


    ax.set_title(
        f"Fiber infrastructure density in Oregon (H3 res {resolution})",
        fontsize=13,
        weight="bold"
    )


    #plt.tight_layout()
    plt.show()
```

```python
df = load_fiber()
```

```python
plot_fiber_density(df, resolution=8)
plot_fiber_density(df, resolution=6)
plot_fiber_density(df, resolution=7)
```
