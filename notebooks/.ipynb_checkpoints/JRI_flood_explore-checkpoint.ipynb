```python
import rioxarray as rxr
import matplotlib.pyplot as plt

# Load raster
flood = rxr.open_rasterio(
    "/home/server/pi/homes/queenl/projects/climate-risk-dc/data/JRC_flood/glofas_rp100_oregon.tif"
)

# Remove the single band dimension
flood_depth = flood.squeeze()

flood_depth.sel(x=slice(-130,-125),y=slice(35,30)).plot()
```

```python
print(flood_depth)
print(flood_depth.x.min().values, flood_depth.x.max().values)
print(flood_depth.y.min().values, flood_depth.y.max().values)
```
